```python
from sklearn.model_selection import train_test_split
X=df.drop(columns=["Class","Hour"])
Y=df["Class"]

X_train,X_temp,Y_train,Y_temp=train_test_split(X,Y,test_size=0.3,random_state=42,stratify=Y)
x_val,x_test,y_val,y_test=train_test_split(X_temp,Y_temp,test_size=0.5,random_state=42,stratify=Y_temp)
```

```python
print("train size :", X_train.shape)
print("test size :", x_test.shape)
print("validation size :", x_val.shape)
```

```python
print("Train:")
print(Y_train.value_counts())

print("\nValidation:")
print(y_val.value_counts())

print("\nTest:")
print(y_test.value_counts())
```

```python
model =Pipeline([
    ("scaler",StandardScaler()),
    ("model",LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

model.fit(X_train,Y_train)

val_probability=model.predict_proba(x_val)[:, 1]

import numpy as np
from sklearn.metrics import precision_recall_curve

precision,recall,threshhold=precision_recall_curve(y_val,val_probability)

f1_scores=2 * (precision * recall) / (precision + recall + 1e-10)
max_fit=np.argmax(f1_scores)


print("Best Threshold:", threshhold[max_fit])
print("Best F1:", f1_scores[max_fit
])
print("Precision:", precision[max_fit])
print("Recall:", recall[max_fit])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    precision_recall_curve,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score
)
import numpy as np


# 1. Completely fresh NORMAL Logistic Regression
final_lr = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# 2. Train ONLY on training set
final_lr.fit(x_train, y_train)


# 3. Validation probabilities
final_val_prob = final_lr.predict_proba(x_val)[:, 1]


# 4. Find threshold ONLY on validation set
val_precision, val_recall, val_thresholds = precision_recall_curve(
    y_val,
    final_val_prob
)

val_f1 = (
    2 * val_precision * val_recall /
    (val_precision + val_recall + 1e-10)
)

best_idx = np.argmax(val_f1)

final_threshold = val_thresholds[best_idx]

print("VALIDATION")
print("Threshold :", final_threshold)
print("Precision :", val_precision[best_idx])
print("Recall    :", val_recall[best_idx])
print("F1        :", val_f1[best_idx])


# 5. Test probabilities FROM THE SAME MODEL
final_test_prob = final_lr.predict_proba(x_test)[:, 1]

final_test_pred = (
    final_test_prob >= final_threshold
).astype(int)


# 6. Final evaluation
print("\nTEST CONFUSION MATRIX")
print(confusion_matrix(y_test, final_test_pred))

print("\nTEST CLASSIFICATION REPORT")
print(
    classification_report(
        y_test,
        final_test_pred,
        digits=4
    )
)

print(
    "\nROC-AUC:",
    roc_auc_score(y_test, final_test_prob)
)

print(
    "PR-AUC:",
    average_precision_score(y_test, final_test_prob)
)
```

## Model 3 - XGBoost

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42
)

xgb_model.fit(x_train, y_train)

xgb_val_prob = xgb_model.predict_proba(x_val)[:, 1]


from sklearn.metrics import precision_recall_curve
import numpy as np

xgb_precision, xgb_recall, xgb_thresholds = precision_recall_curve(
    y_val,
    xgb_val_prob
)

xgb_f1 = (
    2 * xgb_precision * xgb_recall /
    (xgb_precision + xgb_recall + 1e-10)
)

xgb_best_idx = np.argmax(xgb_f1)

xgb_best_threshold = xgb_thresholds[xgb_best_idx]

print("XGB Best Threshold:", xgb_best_threshold)
print("XGB Precision:", xgb_precision[xgb_best_idx])
print("XGB Recall:", xgb_recall[xgb_best_idx])
print("XGB F1:", xgb_f1[xgb_best_idx])
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score
)


xgb_test_prob = xgb_model.predict_proba(x_test)[:, 1]


xgb_test_pred = (
    xgb_test_prob >= xgb_best_threshold
).astype(int)

print("XGBoost Test Confusion Matrix:")
print(confusion_matrix(y_test, xgb_test_pred))

print("XGBoost Test Classification Report:")
print(
    classification_report(
        y_test,
        xgb_test_pred,
        digits=4
    )
)

print(
    "XGB Test ROC-AUC:",
    roc_auc_score(y_test, xgb_test_prob)
)

print(
    "XGB Test PR-AUC:",
    average_precision_score(y_test, xgb_test_prob)
)
```

```python
from xgboost import XGBClassifier

xgb_weighted_model = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric="logloss",
    random_state=42
)

xgb_weighted_model.fit(x_train, y_train)

xgb_weighted_val_prob = xgb_weighted_model.predict_proba(x_val)[:, 1]

from sklearn.metrics import precision_recall_curve
import numpy as np

w_precision, w_recall, w_thresholds = precision_recall_curve(
    y_val,
    xgb_weighted_val_prob
)

w_f1 = (
    2 * w_precision * w_recall /
    (w_precision + w_recall + 1e-10)
)

w_best_idx = np.argmax(w_f1)

w_best_threshold = w_thresholds[w_best_idx]

print("Weighted XGB Threshold:", w_best_threshold)
print("Weighted XGB Precision:", w_precision[w_best_idx])
print("Weighted XGB Recall:", w_recall[w_best_idx])
print("Weighted XGB F1:", w_f1[w_best_idx])
```

```python
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score
)

weighted_test_prob = xgb_weighted_model.predict_proba(x_test)[:, 1]

weighted_test_pred = (
    weighted_test_prob >= w_best_threshold
).astype(int)

print("Weighted XGB Test Confusion Matrix:")
print(confusion_matrix(y_test, weighted_test_pred))

print("Weighted XGB Test Classification Report:")
print(
    classification_report(
        y_test,
        weighted_test_pred,
        digits=4
    )
)

print(
    "\nWeighted XGB ROC-AUC:",
    roc_auc_score(y_test, weighted_test_prob)
)

print(
    "Weighted XGB PR-AUC:",
    average_precision_score(
        y_test,
        weighted_test_prob
    )
)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier

base_xgb = XGBClassifier(
    eval_metric="logloss",
    random_state=42
)

param_dist = {
    "n_estimators": [200, 300, 400, 500],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.01, 0.03, 0.05, 0.1],
    "subsample": [0.7, 0.8, 0.9, 1.0],
    "colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "min_child_weight": [1, 3, 5],
    "gamma": [0, 0.1, 0.3]
}


random_search = RandomizedSearchCV(
    estimator=base_xgb,
    param_distributions=param_dist,
    n_iter=20,
    scoring="average_precision",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=-1
)

random_search.fit(x_train, y_train)

print("Best Parameters:")
print(random_search.best_params_)

print("\nBest CV Score:")
print(random_search.best_score_)
```

```python
tuned_xgb_model = random_search.best_estimator_
tuned_val_prob = tuned_xgb_model.predict_proba(x_val)[:, 1]
from sklearn.metrics import precision_recall_curve
import numpy as np

tuned_precision, tuned_recall, tuned_thresholds = precision_recall_curve(
    y_val,
    tuned_val_prob
)

tuned_f1 = (
    2 * tuned_precision * tuned_recall /
    (tuned_precision + tuned_recall + 1e-10)
)

tuned_best_idx = np.argmax(tuned_f1)

tuned_best_threshold = tuned_thresholds[tuned_best_idx]

print("Tuned XGB Threshold:", tuned_best_threshold)
print("Tuned XGB Precision:", tuned_precision[tuned_best_idx])
print("Tuned XGB Recall:", tuned_recall[tuned_best_idx])
print("Tuned XGB F1:", tuned_f1[tuned_best_idx])
```
